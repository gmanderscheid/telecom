# Partie 3 — Filtrer avec des masques

Un **masque** est une série de `True` / `False`, une valeur par ligne. Placé entre crochets, il ne garde que les lignes à `True`.

```python
masque = df["age_client"] >= 60     # une série de booléens
df[masque]                          # les lignes où c'est True
masque.sum()                        # le nombre de True (True compte pour 1)
```

**Aide-mémoire**

| Besoin | Code |
|---|---|
| ET | `(cond1) & (cond2)` |
| OU | `(cond1) \| (cond2)` |
| NON | `~cond` |
| Appartenance à une liste | `df["col"].isin(["a", "b"])` |
| Entre deux bornes (incluses) | `df["col"].between(10, 20)` |
| Valeur manquante / présente | `df["col"].isna()`, `df["col"].notna()` |
| Texte | `df["col"].str.startswith("L")`, `df["col"].str.contains("web")` |
| Lignes + colonnes | `df.loc[masque, ["col1", "col2"]]` |

Avec pandas on écrit `&` et `|` (pas `and` / `or`), et chaque condition doit être **entre parenthèses**.

# Partie 4 — Créer des colonnes, `apply`


**Aide-mémoire**

```python
df["total"] = df["a"] * df["b"]               # calcul entre colonnes (rapide, à privilégier)
df["mois"] = df["date"].dt.month              # extraire une partie d'une date

def ma_fonction(valeur):                      # une fonction qui traite UNE valeur…
    return valeur * 2
df["double"] = df["a"].apply(ma_fonction)     # …appliquée à chaque valeur de la colonne

df["double"] = df["a"].apply(lambda x: x * 2) # la même chose avec une fonction anonyme

def ma_regle(ligne):                          # une fonction qui reçoit UNE LIGNE entière
    return ligne["a"] + ligne["b"]
df["somme"] = df.apply(ma_regle, axis=1)      # axis=1 : on parcourt les lignes
```

`apply` est l'outil à tout faire : il sert quand la règle est trop particulière pour s'écrire comme un simple calcul entre colonnes.

### Exercice 4.5 — `apply` ligne par ligne : les frais de livraison

La règle dépend de **deux** colonnes, il faut donc travailler sur la ligne entière (`axis=1`) :

- canal `Magasin` : 0 €
- sinon, montant de 60 € ou plus : 0 € (livraison offerte)
- sinon : 4.90 €

Crée la colonne `frais_livraison` et affiche le total des frais encaissés.

````python
def frais_livraison(ligne):
    if ligne["canal"] == "Magasin":
        return 0.0
    elif ligne["montant"] >= 60:
        return 0.0
    else:
        return 4.90


df["frais_livraison"] = df.apply(frais_livraison, axis=1)

print(f"Total des frais de livraison : {round(df["frais_livraison"].sum(), 2)}€")
df["frais_livraison"].value_counts()
````

Crée la même colonne avec np.where puis vérifie la différence de performance entre les deux avec %timeit 

````python
df["frais_livraison"] = np.where((df["canal"] == "Magasin") | (df["montant"] >= 60), 0.0, 4.90)
````

````python
%timeit df.apply(frais_livraison, axis=1)
%timeit np.where((df["canal"] == "Magasin") | (df["montant"] >= 60), 0.0, 4.90)
````

### Exercice 4.6 — `apply` et valeurs manquantes : la satisfaction

Crée la colonne `satisfaction` à partir de `note_client` :

- note manquante → `"Sans avis"`
- note de 4 ou 5 → `"Satisfait"`
- note de 3 → `"Neutre"`
- note de 1 ou 2 → `"Mécontent"`

`pd.isna(valeur)` teste si une valeur est manquante. Fais ce test **en premier** : une comparaison avec une valeur manquante renvoie toujours `False`.

````python
def satisfaction(note):
    if pd.isna(note):
        return "Sans avis"
    elif note >= 4:
        return "Satisfait"
    elif note == 3:
        return "Neutre"
    else:
        return "Mécontent"

df["satisfaction"] = df["note_client"].apply(satisfaction)

df["satisfaction"].value_counts()
````

# Partie 5 — Agréger avec `groupby`

`groupby` répond à toutes les questions du type « … **par** … » : le chiffre d'affaires *par* catégorie, la note moyenne *par* canal. Il procède en trois temps : **découper** le tableau en groupes, **calculer** quelque chose dans chaque groupe, **rassembler** les résultats.

**Aide-mémoire**

```python
df.groupby("categorie")["montant"].sum()                    # une colonne, un calcul
df.groupby("categorie")["montant"].agg(["sum", "mean"])     # une colonne, plusieurs calculs

df.groupby("categorie").agg(                                # plusieurs colonnes, résultats nommés
    nb_commandes=("commande_id", "count"),
    ca=("montant", "sum"),
)

df.groupby(["categorie", "canal"])["montant"].sum()         # deux niveaux de regroupement
```

Utile ensuite : `.sort_values(ascending=False)` pour trier, `.head(5)` pour un top 5, `.round(2)` pour arrondir, `.reset_index()` pour retrouver un DataFrame classique (indispensable pour Plotly).

# Partie 6 — Visualiser avec Plotly

Plotly Express crée un graphique interactif en une ligne : survole les points, zoome, clique sur la légende pour masquer une série.

```python
import plotly.express as px

fig = px.bar(tableau, x="colonne_x", y="colonne_y", color="colonne_couleur", title="Mon titre")
fig.show()
```

**Aide-mémoire**

| Graphique | Fonction | Pour montrer |
|---|---|---|
| Histogramme | `px.histogram(df, x=…, nbins=…)` | la distribution d'une variable numérique |
| Barres | `px.bar(tableau, x=…, y=…)` | une valeur par catégorie |
| Barres groupées | `px.bar(…, color=…, barmode="group")` | une valeur par catégorie et sous-catégorie |
| Nuage de points | `px.scatter(tableau, x=…, y=…, color=…, size=…)` | le lien entre deux variables numériques |
| Courbe | `px.line(tableau, x=…, y=…, markers=True)` | une évolution dans le temps |
| Boîte à moustaches | `px.box(df, x=…, y=…)` | comparer des distributions |

Options communes : `title="…"`, `labels={"colonne": "Libellé affiché"}`, `hover_data=["colonne"]`.

⚠️ Plotly attend un **DataFrame avec des colonnes**. Après un `groupby`, pense à `.reset_index()`.
