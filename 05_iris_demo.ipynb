```python
from sklearn.base import BaseEstimator, TransformerMixin
```

```python
class SigmaClipper(BaseEstimator, TransformerMixin):
    def __init__(self, n_sigma=3.0):
        self.n_sigma = n_sigma 

    def fit(self, X, y=None):
        X_arr = np.asarray(X, dtype=float)
        if hasattr(X, "columns"):
            self.feature_names_in_ = np.asarray(X.columns, dtype=object)
        self.n_features_in_ = X_arr.shape[1]
        self.mean_ = X_arr.mean(axis=0)
        self.std_ = X_arr.std(axis=0)
        self.lower_ = self.mean_ - self.n_sigma * self.std_
        self.upper_ = self.mean_ + self.n_sigma * self.std_
        return self

    def transform(self, X):
        X_arr = np.asarray(X, dtype=float)
        if X_arr.shape[1] != self.n_features_in_:
            raise ValueError(
                f"{X_arr.shape[1]} colonnes reçues, {self.n_features_in_} attendues"
            )
        return np.clip(X_arr, self.lower_, self.upper_)

    def get_feature_names_out(self, input_features=None):
            if input_features is not None:
                return np.asarray(input_features, dtype=object)
            if hasattr(self, "feature_names_in_"):
                return self.feature_names_in_
            return np.asarray([f"x{i}" for i in range(self.n_features_in_)], dtype=object)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=41) # fixer le random state permet la reproductibilité de l'expérience

quantitative_columns = X.select_dtypes(include="float").columns
qualitative_columns = X.select_dtypes(exclude="float").columns

quantitative_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="mean")),
    ("standard", StandardScaler()),
    ("clipper", SigmaClipper(n_sigma=3.0))
])

qualitative_transfomer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(drop="first"))
])

column_transformer = ColumnTransformer([
    ("quantitative", quantitative_transformer, quantitative_columns),
    ("qualitative", qualitative_transfomer, qualitative_columns)
])

column_transformer.fit(X_train)
X_train = column_transformer.transform(X_train)
X_test = column_transformer.transform(X_test)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=41) # fixer le random state permet la reproductibilité de l'expérience

quantitative_columns = X.select_dtypes(include="float").columns
qualitative_columns = X.select_dtypes(exclude="float").columns

quantitative_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="mean")),
    ("standard", StandardScaler()),
    ("clipper", SigmaClipper(n_sigma=3.0))
])

qualitative_transfomer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(drop="first"))
])

column_transformer = ColumnTransformer([
    ("quantitative", quantitative_transformer, quantitative_columns),
    ("qualitative", qualitative_transfomer, qualitative_columns)
])
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
model = Pipeline([
    ("prep", column_transformer),
    ("clf", LogisticRegression(max_iter=1000)),
])

model.fit(X_train, y_train)
model.score(X_test, y_test)

grid = GridSearchCV(
    model,
    param_grid={
        "prep__quantitative__clipper__n_sigma": [1.5, 2.0, 2.5, 3.0, 5.0],
        "clf__C": [0.1, 1.0, 10.0],
    },
    cv=5,
)

grid.fit(X_train, y_train)
```
